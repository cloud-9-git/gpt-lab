## 1. 환경설정

Colab에서는 GitHub 저장소 URL과 GitHub Personal Access Token을 입력해 저장소를 clone하고 `src/`를 import 경로에 추가합니다.
로컬 VS Code에서는 현재 폴더를 프로젝트 루트로 보고 실행합니다.

```python
# Colab: 이 셀을 가장 먼저 실행하세요.
import os
import subprocess
import sys
from pathlib import Path


def normalize_github_url(url: str) -> str:
    """Colab 입력값을 git clone에 사용할 수 있는 https URL로 정리합니다."""
    url = url.strip()
    if not url:
        raise ValueError("GitHub 저장소 URL을 입력해야 합니다.")
    if url.startswith("github.com/"):
        url = "https://" + url
    if not url.startswith("https://"):
        raise ValueError("저장소 URL은 https://github.com/... 또는 github.com/... 형식이어야 합니다.")
    url = url.rstrip("/")
    if not url.endswith(".git"):
        url += ".git"
    return url


if "google.colab" in sys.modules:
    from getpass import getpass

    repo_url = normalize_github_url(input("GitHub 저장소 URL (예: github.com/USERNAME/gpt-lab.git): "))
    token = getpass("GitHub Personal Access Token (Private 저장소인 경우 입력, 공개 저장소면 Enter): ").strip()
    clone_url = repo_url.replace("https://", f"https://{token}@") if token else repo_url
    repo_name = Path(repo_url[:-4]).name if repo_url.endswith(".git") else Path(repo_url).name
    repo_dir = Path("/content") / repo_name

    if not repo_dir.exists():
        subprocess.run(["git", "clone", clone_url, str(repo_dir)], check=True)
        subprocess.run(["git", "remote", "set-url", "origin", repo_url], cwd=repo_dir, check=True)
    else:
        print(f"이미 clone된 저장소를 사용합니다: {repo_dir}")

    os.chdir(repo_dir)
else:
    repo_dir = Path(".").resolve()

sys.path.insert(0, str(repo_dir / "src"))
print(f"Repo: {repo_dir}")
```

```python
# 단계별 테스트 실행 helper
import subprocess
import sys


def run_pytest(target: str):
    cmd = [sys.executable, "-m", "pytest", target, "-v"]
    print("실행 명령:", " ".join(cmd))
    result = subprocess.run(cmd, cwd=str(repo_dir), text=True, capture_output=True)
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
    if result.returncode != 0:
        print("\n아직 통과하지 못한 테스트가 있습니다. 해당 단계의 TODO를 먼저 구현하세요.")
    else:
        print("\n선택한 테스트를 통과했습니다.")
    return result.returncode
```

## 7.5 사전학습 프리셋 실행과 로그 확인

여기에는 테스트 코드를 복사하지 않습니다. 프리셋으로 사전학습 실험을 실행하고, 실행 중 생성되는 epoch 로그를 표와 그래프로 확인합니다.

- 실행: `PRESET` 값만 바꿔 사전학습 실험을 시작합니다.
- 확인: `runs/pretrain_logs/<run-name>/epoch_metrics.csv`를 읽어 epoch별 loss를 확인합니다.
- 보고서: 이 표와 그래프를 바탕으로 실험 결과를 정리합니다.

```python
# 사전학습 프리셋 실행
import subprocess
import sys

PRESETS = [
    "smoke",
    "baseline",
    "batch_2", "batch_4", "batch_8", "batch_16",
    "drop_0_0", "drop_0_1", "drop_0_2",
    "lr_1e_4", "lr_3e_4", "lr_5e_4",
    "ctx_64", "ctx_128",
    "layers_1", "layers_2", "layers_4",
    "emb_64", "emb_128", "emb_192",
]

PRESET = "baseline"
RUN_ALL_PRESETS = False

selected_presets = PRESETS if RUN_ALL_PRESETS else [PRESET]

for preset in selected_presets:
    if preset not in PRESETS:
        raise ValueError(f"알 수 없는 프리셋입니다: {preset}")
    cmd = [
        sys.executable, "scripts/pretrain_local.py",
        "--preset", preset,
        "--no-plot",
        "--train-text", str(LM_TRAIN_PATH),
        "--val-text", str(LM_VAL_PATH),
    ]
    print("=" * 80)
    print("프리셋:", preset)
    print("실행 명령:", " ".join(cmd))
    result = subprocess.run(cmd, cwd=str(repo_dir), text=True, capture_output=True)
    print(result.stdout)
    if result.stderr:
        print(result.stderr)
    if result.returncode != 0:
        print(f"사전학습 프리셋 실패: {preset}")
        break
    else:
        print(f"사전학습 프리셋 완료: {preset}")
```

### 7.5.1 실험 로그 확인

위 프리셋 실행 셀이 만든 최신 로그를 읽어 epoch별 결과를 확인합니다. 특정 실험을 보려면 `LOG_RUN_NAME`에 로그 폴더 이름을 넣으세요.

```python
# 최신 사전학습 실험 로그 확인
import csv
import json
from pathlib import Path

LOG_RUN_NAME = None  # 예: "baseline-5ep". None이면 최신 로그를 자동 선택합니다.
log_root = repo_dir / "runs" / "pretrain_logs"

if LOG_RUN_NAME:
    selected_run_dir = log_root / LOG_RUN_NAME
else:
    run_dirs = [path for path in log_root.iterdir() if path.is_dir()] if log_root.exists() else []
    selected_run_dir = max(run_dirs, key=lambda path: path.stat().st_mtime) if run_dirs else None

metrics = []
if selected_run_dir is None:
    print("아직 실험 로그가 없습니다. 위의 사전학습 프리셋 실행 셀을 먼저 실행하세요.")
else:
    config_path = selected_run_dir / "config.json"
    metrics_path = selected_run_dir / "epoch_metrics.csv"
    print("선택한 로그:", selected_run_dir)
    if config_path.exists():
        config = json.loads(config_path.read_text(encoding="utf-8"))
        print("프리셋:", config.get("preset"))
        print("장치:", config.get("device"))
        print("모델 설정:", config.get("model_config"))
    if metrics_path.exists():
        with metrics_path.open("r", encoding="utf-8") as f:
            metrics = list(csv.DictReader(f))
        columns = ["epoch", "global_step", "epoch_train_loss", "train_eval_loss", "val_loss", "elapsed_sec"]
        print(" | ".join(columns))
        print("-" * 88)
        for row in metrics:
            print(
                f"{row['epoch']:>5} | {row['global_step']:>11} | "
                f"{float(row['epoch_train_loss']):>16.4f} | "
                f"{float(row['train_eval_loss']):>15.4f} | "
                f"{float(row['val_loss']):>8.4f} | "
                f"{float(row['elapsed_sec']):>11.2f}"
            )
    else:
        print("epoch_metrics.csv가 아직 없습니다. 학습이 최소 1 epoch 끝나야 생성됩니다.")
```

```python
# epoch별 loss 그래프
import matplotlib.pyplot as plt

if not metrics:
    print("그래프로 그릴 로그가 없습니다. 위 로그 확인 셀을 먼저 실행하세요.")
else:
    epochs = [int(row["epoch"]) for row in metrics]
    epoch_train_losses = [float(row["epoch_train_loss"]) for row in metrics]
    train_eval_losses = [float(row["train_eval_loss"]) for row in metrics]
    val_losses = [float(row["val_loss"]) for row in metrics]

    plt.figure(figsize=(8, 4))
    plt.plot(epochs, epoch_train_losses, marker="o", label="Epoch Train Loss")
    plt.plot(epochs, train_eval_losses, marker="o", label="Train Eval Loss")
    plt.plot(epochs, val_losses, marker="o", label="Val Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"Pretraining Loss: {selected_run_dir.name}")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()
```
